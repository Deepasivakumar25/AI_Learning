```python
from transformers import pipeline
from pypdf import PdfReader
```

```python
chatbot = pipeline(
    "text-generation",
    model="microsoft/Phi-3-mini-4k-instruct"
)
```

```python
reader = PdfReader("Visit-visa-checklist-Nov-2022.pdf")

pdf_text = ""

for page in reader.pages:
    pdf_text += page.extract_text()
```

```python
while True:

    question = input("\nAsk a question about the PDF (type 'exit' to quit): ")

    if question.lower() == "exit":
        break

    prompt = f"""You are a precise document Q&A assistant.

Rules:
1. Answer using ONLY the text inside the <document> tags below.
2. If the answer is not explicitly stated or clearly inferable from the document, respond exactly with: "I couldn't find that information in the PDF."
3. When you answer, briefly quote or paraphrase the exact supporting sentence from the document.
4. Never use outside knowledge, never guess, never fabricate details.

<document>
{pdf_text}
</document>

<question>
{question}
</question>

Answer:"""

    response = chatbot(
        prompt,
        max_new_tokens=150,
        do_sample=False
    )

    answer = response[0]["generated_text"]

    # Print only the answer after "Answer:"
    print("\nBot:\n")

    if "Answer:" in answer:
        print(answer.split("Answer:")[-1].strip())
    else:
        print(answer)
```
